```python
from PIL import Image
import torchvision.transforms as transforms
import torch
import gradio as gr
```

```python
# Function to preprocess images
def load_image(image):
    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Lambda(lambda x: x.mul(255))
    ])
    return transform(image).unsqueeze(0)
```

```python
# Dummy style transfer function (replace with a real model later)
def apply_style(input_image, style_image):
    return input_image
```

```python
# Gradio interface
def style_transfer(input_img, style_img):
    result = apply_style(input_img, style_img)
    return result

demo = gr.Interface(
    fn=style_transfer,
    inputs=[gr.Image(type="pil"), gr.Image(type="pil")],
    outputs=gr.Image(type="pil"),
    title="Generative AI Style Transfer"
)

demo.launch()
```
